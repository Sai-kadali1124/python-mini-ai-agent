```python
!pip install -q ddgs
from ddgs import DDGS #ddgs seahc and DDGS as search functionality used in ddgs to find
import re # regular expression and find the patterns in the test
#Tool calculator is executed
def calculator_tool(number1,number2):
  result=number1+number2
  print("Calculator tool is Executed: ")
  print("Result: ",result)
#Tool search is executed here
def search_tool(query): # query is user wt he want to search
  print("seaching tool is Executed!")
  print("Seaching",query)
  with DDGS() as ddgs: # with-> handle resourse properly and manage and search session done after clean
    results=ddgs.text(query,max_results=3) #ddgs.text() -> text based web search results
    for result in results:
      print("\nTitle:",result["title"])
      print("Link: ",result["href"])
      print("Body: ",result["body"])
def email_tool(recipient,message):
  print("Email tool is exwcuted: ")
  print("To: ",recipient)
  print("Message is : ",message)
  print("Demo of an email not real email")
# understand the route of the request
def mini_agent(user_request):
  request=user_request.lower()
  if "calculator" in request:
    numbers=re.findall(r"\d+",request) # text in find all numbers to find ex 10 20 10+20=30
    if len(numbers)>=2:
      number1=int(numbers[0])
      number2=int(numbers[1])
      calculator_tool(number1,number2)
    else:
      print("provide the two numbers.")
  elif "search" in request:
    query=request.replace("search","",1).strip()# search first  remove and next empty string for store remanining and 1 for remove only first occurence
    if query:
      search_tool(query) # pass query to the search tool
    else:
      print("Please provide something to search for (e.g., 'search Python tutorials').")
  elif "email" in request:
    email_tool(
        "friend",
        "Hello This is a demo email broo"
    )
  else:
    print("sorry, I dont know which tool you are using..")
user_request=input("ask your agent: ")
mini_agent(user_request) # here we are passig a request to the mini agent
```
